```python
bakery = {
    "name": "Sweet Crumbs",
    "city": "Lahore",
    "hours": "9am to 10 pm",
    "delivery": True,
}

print(bakery["name"])
print(bakery.get("phone", "not listed"))
print("city" in bakery)
print(len(bakery))

bakery["phone"] = "0300-1222333" # to add a new key
bakery["hours"] = "10 am to 11 pm" # to update a value
bakery.update({"city": "Karachi", "delivery": False})
del bakery["delivery"]
hours = bakery.pop("hours")
```

```python
print(list(bakery.keys()))
print(list(bakery.values()))
```

```python
business = {
    "name": "Sweet Crumbs",
    "prices": {"chocolate cake": 1500, "cupcake": 250, "brownie": 300},
    "best_sellers": ["chocolate cake", "brownie"],
}

print(business["prices"]["cupcake"])      # 250
print(business["best_sellers"][0])        # chocolate cake
business["prices"]["donut"] = 200         # add to the inner dictionary
print(len(business["prices"]))            # 4
```
